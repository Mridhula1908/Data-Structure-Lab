```
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def push(self, data):
        ptr1 = Node(data)
        temp = self.head
        ptr1.next = self.head  # New node points to current head

        if self.head is not None:
            # Find the last node
            while temp.next != self.head:
                temp = temp.next
            temp.next = ptr1  # Last node points to the new node (which is now the head)
        else:  # If list is empty
            ptr1.next = ptr1  # Node points to itself

        self.head = ptr1  # New node becomes the head

    def printList(self):
        temp = self.head
        if self.head is not None:
            print("Contents of circular Linked List:", end=" ")
            while True:
                print(temp.data, end=" ")  # Print data on one line, separated by space
                temp = temp.next
                if temp == self.head:
                    break
            print()  # Add a newline at the end of the list display
        else:
            print("List is empty")

# Example Usage
cllist = CircularLinkedList()
cllist.push(12)
cllist.push(56)
cllist.push(2)
cllist.push(11)
cllist.printList()
```
