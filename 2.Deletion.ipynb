```
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        # Initialize head and tail to None for an empty list
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head  # For a circular list, first node points to itself
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head is None:
            print("List is empty, cannot delete from end.")
            return

        if self.head == self.tail:  # Only one node in the list
            self.head = None
            self.tail = None
        else:
            current = self.head
            while current.next != self.tail:
                current = current.next
            current.next = self.head # New tail's next points to head
            self.tail = current     # Update tail to the new last node

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        print("Nodes of the circular linked list:", end=" ")
        while True:
            print(current.data, end=" ")
            current = current.next
            if current == self.head:
                break
        print()

# Example Usage
cl = CircularLinkedList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

print("Original List:")
cl.display()

print("\nDeleting nodes from end:")
# Loop until the list is empty
while cl.head is not None:
    cl.deleteEnd()
    print("Updated List (after deleteEnd):")
    cl.display()

# Adding some elements back to test deletion again if needed
# cl.add(5)
# cl.add(6)
# print("\nList after adding back:")
# cl.display()
```
