```
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        # This method is specifically for when the list is empty.
        # If it's not empty, it returns the current last without adding.
        if self.last is not None:
            return self.last

        temp = Node(data)
        self.last = temp
        self.last.next = self.last # Points to itself for a single-node circular list
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next # New node points to the current head
        self.last.next = temp       # Last node points to the new node (making it the new head)
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next # New node points to the current head
        self.last.next = temp       # Current last node points to the new node
        self.last = temp            # New node becomes the new last
        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            print("List is empty, cannot add after.")
            return None

        temp = Node(data)
        p = self.last.next # Start from the head
        found = False
        while True: # Iterate through the circular list
            if p.data == item:
                temp.next = p.next
                p.next = temp
                if p == self.last: # If we added after the last node, new node becomes last
                    self.last = temp
                found = True
                break # Item found and data added
            p = p.next
            if p == self.last.next: # If we've completed a full circle and haven't found item
                break

        if not found:
            print(item, "not present in the list")
        return self.last

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next # Start from the head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.last.next: # If we've circled back to the head
                break
        print() # Newline after printing all elements


if __name__ == '__main__':
    llist = CircularLinkedList()
    llist.addToEmpty(6)
    llist.addBegin(4)
    llist.addBegin(2)
    llist.addEnd(8)
    llist.addEnd(12)
    llist.addAfter(10, 8) # Add 10 after 8
    llist.traverse()
```
